```python
import sys, torch, transformers, accelerate
print('python      :', sys.version.split()[0])
print('torch       :', torch.__version__)
print('transformers:', transformers.__version__)
print('accelerate  :', accelerate.__version__)
print('cuda        :', torch.cuda.is_available())
if torch.cuda.is_available():
    p = torch.cuda.get_device_properties(0)
    print('GPU         :', p.name, f'{p.total_memory/1e9:.1f} GB VRAM')
```

```python
# hello world: gera uma imagem sintética e carrega com PIL (troque por uma foto sua em data/)
from pathlib import Path
from PIL import Image, ImageDraw

Path('../data').mkdir(exist_ok=True)
img_path = Path('../data/teste.png')
if not img_path.exists():
    im = Image.new('RGB', (384, 384), 'white')
    d = ImageDraw.Draw(im)
    d.rectangle([80, 80, 300, 300], outline='black', width=6)
    d.ellipse([140, 140, 240, 240], fill='red')
    im.save(img_path)

img = Image.open(img_path).convert('RGB')
print('imagem carregada:', img.size, img.mode)
img
```
